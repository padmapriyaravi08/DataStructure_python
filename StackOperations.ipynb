```python
class Stack:
    def __init__(self, capacity):
        self.capacity = capacity
        self.items = []

    def push(self, item):
        if self.isFull():
            print("Stack is full! Cannot push.")
        else:
            self.items.append(item)
            print(f"Pushed: {item}")

    def pop(self):
        if self.isEmpty():
            print("Stack is empty! Cannot pop.")
        else:
            popped_item = self.items.pop()
            print(f"Popped: {popped_item}")

    def peek(self):
        if self.isEmpty():
            print("Stack is empty! Nothing to peek.")
        else:
            print(f"Top element: {self.items[-1]}")

    def isEmpty(self):
        return len(self.items) == 0

    def isFull(self):
        return len(self.items) == self.capacity

    def display(self):
        if self.isEmpty():
            print("Stack is empty.")
        else:
            print("Stack contents (top → bottom):", list(reversed(self.items)))


# Main program
try:
    capacity = int(input("Enter stack capacity: "))
    if capacity <= 0:
        raise ValueError("Capacity must be a positive integer.")
except ValueError as e:
    print(f"Invalid input: {e}")
    exit(1)

stack = Stack(capacity)

while True:
    print("\n--- Stack Operations Menu ---")
    print("1. Push")
    print("2. Pop")
    print("3. Peek")
    print("4. Check if Empty")
    print("5. Check if Full")
    print("6. Display Stack")
    print("7. Exit")

    choice = input("Enter your choice (1-7): ").strip()

    if choice == '1':
        item = input("Enter item to push: ")
        stack.push(item)
    elif choice == '2':
        stack.pop()
    elif choice == '3':
        stack.peek()
    elif choice == '4':
        print("Stack is empty." if stack.isEmpty() else "Stack is not empty.")
    elif choice == '5':
        print("Stack is full." if stack.isFull() else "Stack is not full.")
    elif choice == '6':
        stack.display()
    elif choice == '7':
        print("Exiting program.")
        break
    else:
        print("Invalid choice! Please enter a number between 1 and 7.")
```

```python
stack = []
capacity = int(input("Enter stack capacity: "))

while True:
    print("\n1. Push")
    print("2. Pop")
    print("3. Peek")
    print("4. Check if Empty")
    print("5. Check if Full")
    print("6. Display Stack")
    print("7. Exit")

    choice = input("Enter choice (1-7): ")

    if choice == '1':  # Push
        if len(stack) == capacity:
            print("Stack is full!")
        else:
            item = input("Enter item to push: ")
            stack.append(item)
            print(f"Pushed: {item}")

    elif choice == '2':  # Pop
        if not stack:
            print("Stack is empty!")
        else:
            print(f"Popped: {stack.pop()}")

    elif choice == '3':  # Peek
        if not stack:
            print("Stack is empty!")
        else:
            print(f"Top element: {stack[-1]}")

    elif choice == '4':  # isEmpty
        print("Stack is empty." if not stack else "Stack is not empty.")

    elif choice == '5':  # isFull
        print("Stack is full." if len(stack) == capacity else "Stack is not full.")

    elif choice == '6':  # Display
        print("Stack:", list(reversed(stack)) if stack else "Empty")

    elif choice == '7':  # Exit
        print("Exiting...")
        break

    else:
        print("Invalid choice! Please enter 1-7.")
```
